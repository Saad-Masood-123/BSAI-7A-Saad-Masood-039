# Task 3 - Distributions and Relationships

### Dataset: `penguins.csv`

This task looks at distributions and relationships in penguin measurements.

The dataset contains missing values, so we will check them before making charts.

## 1. Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Load the data

```python
data = pd.read_csv("data/penguins.csv")

data.head()
```

## 3. Check missing values

```python
print("Rows:", len(data))
print("Columns:", list(data.columns))

data.isna().sum()
```

### Missing-value decision

The task asks us to find the missing values and state what we do with them.

For the measurement charts, rows without the required measurement values will be removed. For the sex comparison, rows without `sex` will also be removed.

We will keep the original data unchanged and create cleaned copies for each chart.

## 4. Body mass distribution

```python
body_mass = data.dropna(subset=["body_mass_g"])

plt.figure(figsize=(10, 6))
plt.hist(body_mass["body_mass_g"], bins=20)
plt.title("Penguin Body Mass Has More Than One Main Peak")
plt.xlabel("Body mass (g)")
plt.ylabel("Number of penguins")
plt.show()
```

### Why this chart?

A histogram is used because the question is about the **distribution** of one numeric variable.

### What does it show?

The body-mass distribution is not a simple single smooth hump. The different penguin species are likely contributing to the different parts of the distribution.

## 5. Body mass split by species

```python
species_mass = data.dropna(subset=["body_mass_g"])

plt.figure(figsize=(10, 6))

for species in species_mass["species"].unique():
    values = species_mass[species_mass["species"] == species]["body_mass_g"]
    plt.hist(values, bins=15, alpha=0.5, label=species)

plt.title("Species Explain the Different Body-Mass Ranges")
plt.xlabel("Body mass (g)")
plt.ylabel("Number of penguins")
plt.legend()
plt.show()
```

### Why this chart?

The same histogram is split by species so we can compare the distributions of the groups.

The separate groups help explain why the overall body-mass distribution has more than one peak.

## 6. Bill length and bill depth

```python
bill_data = data.dropna(subset=["bill_length_mm", "bill_depth_mm"])

plt.figure(figsize=(10, 6))

for species in bill_data["species"].unique():
    part = bill_data[bill_data["species"] == species]
    plt.scatter(
        part["bill_length_mm"],
        part["bill_depth_mm"],
        label=species,
        alpha=0.7
    )

plt.title("Bill Length and Depth Look Different Within and Across Species")
plt.xlabel("Bill length (mm)")
plt.ylabel("Bill depth (mm)")
plt.legend()
plt.show()
```

### Why this chart?

A scatter plot is the right choice because both variables are numeric and we want to see how they move together.

The overall relationship can look negative when all penguins are combined, while the relationship inside individual species can look positive.

This is an example of **Simpson's paradox**: a relationship seen after combining groups can be different from the relationships seen inside the groups.

## 7. Average body mass by species and sex

```python
sex_data = data.dropna(subset=["body_mass_g", "sex"])

average_mass = sex_data.groupby(["species", "sex"])["body_mass_g"].mean().unstack()

average_mass
```

```python
average_mass.plot(kind="bar", figsize=(10, 6))

plt.title("Average Body Mass Differs by Species and Sex")
plt.xlabel("Species")
plt.ylabel("Average body mass (g)")
plt.xticks(rotation=0)
plt.legend(title="Sex")
plt.show()
```

### Why this chart?

A grouped bar chart makes it easy to compare two categories at the same time: species and sex.

The missing `sex` rows were excluded from this chart because they cannot be placed into either sex group.

### Final takeaway

The penguins task shows why grouping matters. A single overall chart can hide patterns that become visible after separating the data by species.
